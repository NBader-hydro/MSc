```python
import numpy as np
import copy
```

```python
# Chargement des données
bh_lbh_0 = np.loadtxt("selected_boreholes2.csv", delimiter=";", skiprows=1)
all_unit_id = np.loadtxt("all_units.csv", delimiter=";", skiprows=1)
all_unit = np.loadtxt("all_units.txt", delimiter="\t", skiprows=1)
all_quat = np.loadtxt("all_quat.txt", delimiter="\t", skiprows=1, dtype=[('PROFONDEUR_TOIT', np.float_), ('PROFONDEUR_BASE', np.float_), ('ID_UNITE_GEOL', np.float_), ('NPXY', 'U50')])
```

```python
bh_ids = bh_lbh_0[:,0]

bh_ud = copy.deepcopy(all_unit)

fd_dtop, fd_dbot, fd_ud_geol, fd_npxy = [], [], [], []

for i in range(len(all_quat)):

    fd_dtop.append(all_quat[i][0])
    fd_dbot.append(all_quat[i][1])
    fd_ud_geol.append(all_quat[i][2])
    fd_npxy.append(all_quat[i][3])
```

```python
bh_ud = []
ud_ids = []

for j in range(len(all_unit)):
    ud_id_bh = all_unit[j][0]

    if ud_id_bh in bh_ids :
        bh_ud.append(all_unit[j].tolist())
        ud_ids.append([all_unit_id[j][1]])
```

```python
bh_ud2 = []

for g in range(len(bh_ud)):
    bh_ud2.append([float(bh_ud[g][0]), str(bh_ud[g][1]), float(bh_ud[g][2]), float(bh_ud[g][3])])
```

```python
for e in range(len(bh_ud)):
    unit = bh_ud[e][1]
    if unit == 1 or unit == 2 or unit == 3 or unit == 4 or unit == 5 or unit == 38 or unit == 39 :
        bh_ud2[e][1] = 'u3'
    elif unit == 6 or unit == 7 or unit == 8:
        bh_ud2[e][1] = 'u7'
    elif unit == 9 :
        bh_ud2[e][1] = 'u9'
    elif unit == 10 or unit == 11 or unit == 12 or unit == 13 :
        bh_ud2[e][1] = 'u12'
    elif unit == 14 or unit == 15 or unit == 18 :
        bh_ud2[e][1] = 'u14'
```

```python
bh_fd = []
fd_id = []

for j in range(len(all_quat)):
    fd_id_bh = all_quat[j][2]
    fd = all_quat[j][3]
    alt_top = all_quat[j][0]
    alt_bot = all_quat[j][1]

    if fd_id_bh in ud_ids :
        #bh_fd.append(all_quat[j].tolist())
        for k in range(len(all_unit)):
            #print(all_unit_id[k][1], fd_id_bh)
            if all_unit_id[k][1] == fd_id_bh :
                id_adm = all_unit_id[k][0]
                for m in range(len(bh_lbh_0)):
                    if bh_lbh_0[m,0] == id_adm:
                        fd_id.append(id_adm)
                        bh_alt = bh_lbh_0[m,3]
                        top = bh_alt - alt_top
                        bot = bh_alt - alt_bot
        bh_fd.append([id_adm, fd, top, bot])
```

```python
for f in range(len(bh_fd)):
    facies = bh_fd[f][1]
    if facies == '1' or facies == '2' or facies == '1-2':
        bh_fd[f][1] = 'Soil'
    elif 'a' in facies and 'b' not in facies and 'c' not in facies:
        bh_fd[f][1] = 'a'
    elif 'ab' in facies or 'ba' in facies:
        bh_fd[f][1] = 'ab'
    elif 'ac' in facies or 'ca' in facies:
        bh_fd[f][1] = 'ac'
    elif 'b' in facies and 'a' not in facies and 'c' not in facies:
        bh_fd[f][1] = 'b'
    elif 'bc' in facies or 'cb' in facies:
        bh_fd[f][1] = 'bc'
    elif 'c' in facies and 'b' not in facies and 'a' not in facies:
        bh_fd[f][1] = 'c'
    elif 'd' in facies:
        bh_fd[f][1] = 'd'
    elif 'e' in facies:
        bh_fd[f][1] = 'e'
```

```python
for units in bh_ud:
    if units[1] == 14 and units[0] not in fd_id:
        bh_fd.append([units[0],'Molasse',units[2], units[3]])
        
```

```python
# Specify the file path
file_path = "versoix_bh.fd"

# Using "with open" syntax to automatically close the file
with open(file_path, 'w') as file:
        file.write('bh_ID;facies_ID;top;bot\n')
    
        for lists in bh_fd:
            it_a = lists[0]
            it_b = lists[1]
            it_c = lists[2]
            it_d = lists[3]
            file.write(f'{it_a};{it_b};{it_c};{it_d}\n')
```

```python
# Specify the file path
file_path = "versoix_bh.ud"

# Using "with open" syntax to automatically close the file
with open(file_path, 'w') as file:
        file.write('bh_ID;Strat;top;bot\n')
    
        for lists in bh_ud2:
            it_a = lists[0]
            it_b = lists[1]
            it_c = lists[2]
            it_d = lists[3]
            file.write(f'{it_a};{it_b};{it_c};{it_d}\n')
```
